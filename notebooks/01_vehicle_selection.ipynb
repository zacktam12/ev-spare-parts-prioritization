## 1.3 Selection Decision

### Selected: BYD Seagull, BYD Yuan Up, Toyota bZ4X

### Reasoning for each vehicle:

```python
selection_reasoning = pd.DataFrame({
    'vehicle': ['BYD Seagull', 'BYD Yuan Up', 'Toyota bZ4X'],
    'reason_selected': [
        'Strong marketplace visibility (multiple Adey listings + Gezat catalogue). '
        'Two dedicated parts catalogues available (cqswautoparts.com for part numbers, '
        'landinparts.com for pricing). This gives us the best chance of finding '
        'compatibility and pricing evidence for all 10 part categories.',
        
        'Strongest listing volume on Adey Market (1,360+ listings for BYD Yuan Up page). '
        'Listed on Gezat catalogue. Parts available through landinparts.com. '
        'High marketplace visibility suggests significant import volume into Ethiopia.',
        
        'Toyota has established brand recognition and service infrastructure in Ethiopia. '
        'The bZ4X appears on both Adey Market and Gezat. A dedicated parts catalogue '
        'exists at toyota-spare-parts.com. NHTSA complaint data is available for the '
        'bZ4X (US market), providing independent failure evidence.'
    ]
})

selection_reasoning.style.set_properties(**{'text-align': 'left', 'white-space': 'pre-wrap'})
```

### Why the other two were not selected:

```python
rejection_reasoning = pd.DataFrame({
    'vehicle': ['Changan Benben E-Star', 'Volkswagen ID.4'],
    'reason_not_selected': [
        'Weakest evidence among all five candidates. Adey Market has only one listing '
        'labelled "Estar" — the exact model (Benben E-Star) could not be confirmed. '
        'Gezat lists the Changan brand but the specific Benben E-Star model needs '
        'verification. No dedicated parts catalogue was found, which would make '
        'investigating all 10 part categories very difficult.',
        
        'The VW ID.4 has adequate marketplace evidence (Adey listings + Gezat catalogue) '
        'and a parts catalogue at exist.ae. However, Toyota has stronger brand '
        'recognition and existing service infrastructure in Ethiopia, making the bZ4X '
        'a more practical import recommendation. The ID.4 would be a valid alternative '
        'choice — this is a judgement call, not a data gap.'
    ]
})

rejection_reasoning.style.set_properties(**{'text-align': 'left', 'white-space': 'pre-wrap'})
```

## 1.4 Research Scope

With 3 vehicles × 10 part categories, we have **30 combinations** to investigate:

```python
vehicles = ['BYD Seagull', 'BYD Yuan Up', 'Toyota bZ4X']
part_categories = [
    'Brake pads',
    'Brake discs',
    'Cabin-air filter',
    'Wiper blades',
    'Headlight assembly',
    'Tail-light assembly',
    'Side mirror',
    'Bumper',
    'Fender / body panel',
    'Shock absorber'
]

# Generate all 30 combinations
research_matrix = pd.DataFrame(
    [(v, p) for v in vehicles for p in part_categories],
    columns=['vehicle', 'part_category']
)
research_matrix['status'] = 'to_investigate'

print(f"Total combinations to research: {len(research_matrix)}")
print(f"Vehicles: {len(vehicles)}")
print(f"Part categories: {len(part_categories)}")
print()
research_matrix
```
